```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp
from scipy.signal import square
t = np.linspace(0,1,1000)
a1 = 2
a2 = 3
f3 = 5
f4 = 10

x1 = a1*np.sin(2*np.pi*f3*t)
x2 = a2*square(2*np.pi*f4*t)

gain = 3
y1 = gain*x1
y2 = gain*x2
plt.figure(figsize=(10,10))
plt.subplot(2,1,1)
plt.plot(t,y1)
plt.plot(t,x1)

plt.subplot(2,1,2)
plt.plot(t,x2)
plt.plot(t,y2)

plt.show()
```
