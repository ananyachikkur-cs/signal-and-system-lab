```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import square

# AMPLIFICATION OF THE SIGNALS #
t = np.linspace(0,1,1000)
a1 = 2
a2 = 3
f3 = 5
f4 = 10

x1 = a1*np.sin(2*np.pi*f3*t)
x2 = a2*square(2*np.pi*f4*t)

gain = 3
y1 = gain*x1
y2 = gain*x2
plt.figure(figsize=(10,10))
plt.subplot(2,1,1)
plt.plot(t,y1)
plt.plot(t,x1)
plt.xlabel('Time (s)')
plt.ylabel('Amplitude')

plt.subplot(2,1,2)
plt.plot(t,x2)
plt.plot(t,y2)
plt.xlabel('Time (s)')
plt.ylabel('Amplitude')


plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import square

#MULTIPLICATION OF THE TWO SIGNALS#
t = np.linspace(0,1,1000)
a1 = 2
a2 = 3
f3 = 5
f4 = 10

x1 = a1*np.sin(2*np.pi*f3*t)
x11 = a2*np.sin(2*np.pi*f4*t)
x2 = a1*square(2*np.pi*f4*t)
x22 = a2*square(2*np.pi*f3*t)
y1 = x1*x2
y2 = x1*x11
y3 = x2*x22
plt.figure(figsize=(8,15))
plt.subplot(3,1,1)
plt.plot(t,y1,'y')
plt.title('Multiplication of sin and square')
plt.xlabel('Time (s)')
plt.ylabel('Amplitude')
plt.subplot(3,1,2)
plt.plot(t,y2,'g')
plt.title('Multiplication of sin and sin')
plt.xlabel('Time (s)')
plt.ylabel('Amplitude')
plt.subplot(3,1,3)
plt.plot(t,y3,)
plt.title('Multiplication of square and square')
plt.xlabel('Time (s)')
plt.ylabel('Amplitude')

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import square

#MULTIPLICATION OF THE TWO SIGNALS#
t = np.linspace(0,1,1000)
a1 = 2
f3 = 5
f4 = 10

x1 = a1*np.sin(2*np.pi*f3*t)
x2 = a1*square(2*np.pi*f4*t)
dt = t[0] - t[1]
y1 = np.gradient(x1,dt)
y2 = np.gradient(x2,dt)

plt.figure(figsize=(10,10))
plt.subplot(2,1,1)
plt.plot(t,y1)
plt.plot(t,x1)
plt.xlabel('Time (s)')
plt.ylabel('Amplitude')

plt.subplot(2,1,2)
plt.plot(t,x2)
plt.plot(t,y2)
plt.xlabel('Time (s)')
plt.ylabel('Amplitude')

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import square
t = np.linspace(0,1,1000)
a1 = 2
f3 = 5
f4 = 10
dt = t[0] - t[1]
x1 = a1*np.sin(2*np.pi*f3*t)
x2 = a1*square(2*np.pi*f4*t)

y1 =  np.cumsum(x1)*dt
y2 =  np.cumsum(x2)*dt

plt.figure(figsize=(10,10))
plt.subplot(2,1,1)
plt.plot(t,y1)
plt.xlabel('Time (s)')
plt.ylabel('Amplitude')

plt.subplot(2,1,2)
plt.plot(t,y2)
plt.xlabel('Time (s)')
plt.ylabel('Amplitude')

plt.show()
```
