```python
#Even or odd
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal as si
import sympy as sy
t = np.linspace(-2,2,1000)
f = 2
A = 2
x1 = A*np.sin(2*np.pi*t*f)
x2 = A*np.cos(2*np.pi*t*f)
x3 = A*np.sin(-2*np.pi*t*f)
x4 = A*np.cos(-2*np.pi*t*f)
odd = np.allclose(x1,-x3)
even = np.allclose(x2,x4)
print(f"sin(pi*t) is odd? {odd})")
print(f"cos(pi*t) is even ? {even})")

plt.figure(figsize=(10,5))
plt.subplot(2,1,1)
plt.plot(t,x1)
plt.plot(t,x3)
plt.title("Odd")
plt.subplot(2,1,2)
plt.plot(t,x2)
plt.plot(t,x4)
plt.title("Even")
plt.show()
```

```python
#periodic or aperiodic signals

import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(-4, 4, 1000)

periodic = np.sin(4*np.pi*t)
aperiodic = np.exp(-1.5*t)

ac_p = np.correlate(periodic, periodic, mode='full')
ac_ap = np.correlate(aperiodic, aperiodic, mode='full')

lags = np.linspace(-4, 4, len(ac_p))
plt.figure(figsize=(10,7))

plt.subplot(2,2,1)
plt.plot(t, periodic)
plt.title("Periodic Signal")
plt.xlabel("Time (t)")

plt.subplot(2,2,2)
plt.plot(lags, ac_p)
plt.title("Autocorrelation of Periodic Signal")
plt.xlabel("Lag")

plt.subplot(2,2,3)
plt.plot(t, aperiodic)
plt.title("Aperiodic Signal")
plt.xlabel("Time (t)")

plt.subplot(2,2,4)
plt.plot(lags, ac_ap)
plt.title("Autocorrelation of Aperiodic Signal")
plt.xlabel("Lag")

plt.tight_layout()
plt.show()
```
