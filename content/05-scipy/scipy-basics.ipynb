```python
import numpy as np
import matplotlib.pyplot as plt
# unit impulse function
t = np.arange(-10,10)
x = np.zeros(len(t))
x[10] = 1
plt.figure()
plt.stem(t,x)
plt.show()

```

```python
import numpy as np
import matplotlib.pyplot as plt
# unit impulse fuction
t = np.arange(-10,10)
n = np.zeros(len(t))
for i in range(len(t)):
  if t[i]>=0:
    n[i] = 1
plt.figure()
plt.stem(t,n)
plt.show()

```
