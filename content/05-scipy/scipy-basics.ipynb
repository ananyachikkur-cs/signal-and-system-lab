```python
import numpy as np
import matplotlib.pyplot as plt
# unit impulse function
t = np.arange(-10,10)
x = np.zeros(len(t))
x[10] = 1
plt.figure()
plt.stem(t,x)
plt.show()

```

```python
import numpy as np
import matplotlib.pyplot as plt

# Define the time vector and initialize values
t = np.linspace(-10, 10, 100)
n = np.zeros(len(t))
for i in range(len(t)):
  if t[i] >= 0:
    n[i] = 1

# Create a figure for side-by-side subplots
plt.figure(figsize=(12, 5))

# Discrete-Time (DT) step function
plt.subplot(1, 2, 1)
plt.stem(t, n)
plt.title('Unit Step Function (Discrete-Time DT)')
plt.xlabel('t')
plt.ylabel('n')
plt.grid(True)

# Right subplot: Continuous-Time (CT) step function
plt.subplot(1, 2, 2)
plt.plot(t, n)
plt.title('Unit Step Function (Continuous-Time CT)')
plt.xlabel('t')
plt.ylabel('n')
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
t = np.linspace(-10, 10, 100)
n = np.zeros(len(t))
for i in range(len(t)):
  if t[i] >= 0:
    n[i] = t[i]

# Create a figure for side-by-side subplots
plt.figure(figsize=(12, 5))

# Discrete-Time (DT) ramp signal
plt.subplot(1, 2, 1)
plt.stem(t, n)
plt.title('Unit Step Function (Discrete-Time DT)')
plt.xlabel('t')
plt.ylabel('n')
plt.grid(True)

# Continuous-Time (CT) ramp signal
plt.subplot(1, 2, 2)
plt.plot(t, n)
plt.title('Unit Step Function (Continuous-Time CT)')
plt.xlabel('t')
plt.ylabel('n')
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal as s
t = np.linspace(-5,5,1000)
t1 = np.linspace(-5,5,100)
n1 = s.sawtooth(2*np.pi*t1)
n = s.sawtooth(2*np.pi*t)
# Create a figure for side-by-side subplots
plt.figure(figsize=(12, 5))

# Discrete-Time (DT) sawtooth
plt.subplot(1, 2, 1)
plt.stem(t1, n1)
plt.title('sawtooth (Discrete-Time DT)')
plt.xlabel('t')
plt.ylabel('n')
plt.grid(True)

# Continuous-Time (CT) sawtooth
plt.subplot(1, 2, 2)
plt.plot(t, n)
plt.title('sawtooth (Continuous-Time CT)')
plt.xlabel('t')
plt.ylabel('n')
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal as s

#by changing the or by fixing the widht we can make sawtooth into trianglewave

t = np.linspace(-5,5,1000)
t1 = np.linspace(-5,5,100)
n1 = s.sawtooth(2*np.pi*t1,width = 0.35)
n = s.sawtooth(2*np.pi*t, width = 0.35)
# Create a figure for side-by-side subplots
plt.figure(figsize=(12, 5))

# Discrete-Time (DT triangle
plt.subplot(1, 2, 1)
plt.stem(t1, n1)
plt.title('taiangle (Discrete-Time DT)')
plt.xlabel('t')
plt.ylabel('n')
plt.grid(True)

# Continuous-Time (CT) triangle
plt.subplot(1, 2, 2)
plt.plot(t, n)
plt.title('triangle (Continuous-Time CT)')
plt.xlabel('t')
plt.ylabel('n')
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal as s

t = np.linspace(0,2,1000)
t1 = np.linspace(0,2,100)
f = 2
A = 2
n = A*np.sin(2*np.pi*f*t)
n1 = A*np.sin(2*np.pi*f*t1)
plt.figure(figsize=(12, 5))

# Discrete-Time sin signal
plt.subplot(1, 2, 1)
plt.stem(t1, n1)
plt.title('sinusodial (Discrete-Time DT)')
plt.xlabel('t')
plt.ylabel('n')
plt.grid(True)

# Continuous-Time sin signal
plt.subplot(1, 2, 2)
plt.plot(t, n)
plt.title('sinusoidal (Continuous-Time CT)')
plt.xlabel('t')
plt.ylabel('n')
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Define continuous and discrete time vectors
t = np.linspace(-5, 5, 1000)
t1 = np.linspace(-5, 5, 10)

# Calculate signum function values
x = np.sign(t)
x1 = np.sign(t1)

plt.figure(figsize=(12, 5))

# Discrete-Time (DT) signum signal
plt.subplot(1, 2, 1)
plt.stem(t1, x1)
plt.title('Signum (Discrete-Time DT)')
plt.xlabel('t')
plt.ylabel('n')
plt.grid(True)

# Continuous-Time (CT) signum signal
plt.subplot(1, 2, 2)
plt.plot(t, x)
plt.title('Signum (Continuous-Time CT)')
plt.xlabel('t')
plt.ylabel('n')
plt.grid(True)

plt.tight_layout()
plt.show()
```
