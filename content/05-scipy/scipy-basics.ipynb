```python
import numpy as np
import matplotlib.pyplot as plt
# unit impulse function
t = np.arange(-10,10)
x = np.zeros(len(t))
x[10] = 1
plt.figure()
plt.stem(t,x)
plt.show()

```

```python
import numpy as np
import matplotlib.pyplot as plt

# Define the time vector and initialize values
t = np.linspace(-10, 10, 100)
n = np.zeros(len(t))
for i in range(len(t)):
  if t[i] >= 0:
    n[i] = 1

# Create a figure for side-by-side subplots
plt.figure(figsize=(12, 5))

# Discrete-Time (DT) step function
plt.subplot(1, 2, 1)
plt.stem(t, n)
plt.title('Unit Step Function (Discrete-Time DT)')
plt.xlabel('t')
plt.ylabel('n')
plt.grid(True)

# Right subplot: Continuous-Time (CT) step function
plt.subplot(1, 2, 2)
plt.plot(t, n)
plt.title('Unit Step Function (Continuous-Time CT)')
plt.xlabel('t')
plt.ylabel('n')
plt.grid(True)

plt.tight_layout()
plt.show()
```
