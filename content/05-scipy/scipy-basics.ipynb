```python
import numpy as np
import matplotlib.pyplot as plt
import scipy as sp
import sympy as sy

n = np.arange(-10, 11)

x_impulse = np.zeros(len(n))
x_impulse[10] = 1
plt.figure()
plt.stem(n, x_impulse)
plt.title("Unit Impulse Signal (Hardcoded)")
plt.show()

x_step = np.zeros(len(n))
x_step[n>=0] = 1
plt.figure()
plt.stem(n,x_step)
plt.show()

x_step = np.heaviside(n, 0)
plt.figure()
plt.stem(n, x_step)
plt.title("Unit Step Signal")
plt.show()

x_imp_correct = np.where(n == 0, 1, 0)
plt.figure()
plt.stem(n, x_imp_correct)
plt.title("Unit Impulse Signal (Using np.where)")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sy
#ramp function
n =np.arange(-10,11)
x = np.zeros(len(n))
for i in range(len(n)):
      if n[i]>=0: x[i] = n[i]
plt.figure()
plt.stem(n,x)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal
import sympy as sy
#sawtooth ct and dt
t = np.linspace(-10,10,1000)
n = signal.sawtooth(2*np.pi*t)
plt.subplot(2,1,1)
plt.plot(t,n)
plt.show()

t1 = np.linspace(-10,10,50)
n1 = signal.sawtooth(2*np.pi*t1)
plt.subplot(2,1,2)
plt.stem(t1,n1)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal
import sympy as sp

#triangle wave
# by change or giving the widht we can change the sawtooth into trangularwave

t = np.linspace(-10,10,1000)
n = signal.sawtooth(2*np.pi*t, width = 0.6)
plt.subplot(2,1,1)
plt.plot(t,n)
plt.show()

t1 = np.linspace(-10,10,50)
n1 = signal.sawtooth(2*np.pi*t1)
plt.subplot(2,1,2)
plt.stem(t1,n1)
plt.show()
```
